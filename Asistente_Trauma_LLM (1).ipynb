# AI Assistant for Lower-Limb Trauma Patient FAQs

This notebook builds a simple conversational assistant powered by an LLM to answer common questions from patients who suffered road traffic accidents with soft-tissue injury in the lower limbs.

Example questions it can help address:
- Why didn’t the injury show up immediately?
- Why is the skin turning black?
- What are the goals of surgery?

The goal is patient-friendly education: clear, empathetic, plain-language explanations. This is a prototype for learning and testing prompt behavior and does not replace medical assessment.

# Asistente de IA para Pacientes con Trauma en Extremidades Inferiores
Este cuaderno está diseñado para construir un asistente conversacional que responda, con apoyo de un LLM, preguntas frecuentes de pacientes que han sufrido accidentes de tránsito con compromiso de tejidos blandos en miembros inferiores.

**Ejemplos de preguntas que puede responder:**
- ¿Por qué la lesión no se manifestó desde el inicio?
- ¿Por qué la piel toma color negro?
- ¿Qué objetivos tiene la cirugía?

Este cuaderno puede ser conectado más adelante a WhatsApp utilizando una API como Twilio o Meta Cloud.

```python
from getpass import getpass
OPENAI_API_KEY = getpass('Ingresa tu clave de OpenAI: ')
```

```python
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from langchain_community.llms import OpenAI

plantilla = '''Eres un médico experto en trauma de extremidades inferiores. Responde de forma clara, empática y en lenguaje sencillo la siguiente pregunta de un paciente:

Pregunta: {pregunta}

Respuesta:'''

prompt = PromptTemplate(
    input_variables=['pregunta'],
    template=plantilla
)
```

```python
llm = OpenAI(openai_api_key=OPENAI_API_KEY, temperature=0)
cadena = LLMChain(llm=llm, prompt=prompt)
```

```python
# Ejemplo: el paciente pregunta por qué la piel se pone negra tras el accidente
pregunta = '¿Por qué la piel se pone negra después del accidente?'
respuesta = cadena.run(pregunta=pregunta)
print(respuesta)
```

```python
# Puedes cambiar esta pregunta por otra que te interese resolver
pregunta = input('Escribe tu pregunta: ')
respuesta = cadena.run(pregunta=pregunta)
print('\nRespuesta del asistente:\n')
print(respuesta)
```
